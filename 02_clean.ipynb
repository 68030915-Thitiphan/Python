```python
import pandas as pd

file_path = "../data/raw/Game_data.xlsx"

df = pd.read_excel(file_path)

# 2. แปลงข้อมูลในคอลัมน์ Year ให้เป็นตัวเลข (ลบแถวที่เป็น N/A หรือไม่มีข้อมูลออกก่อน)
df = df.dropna(subset=["Year"])
df["Year"] = df["Year"].astype(int)

# 3. กรองเงื่อนไข Publisher เฉพาะ Nintendo, Ubisoft, Electronic Arts
# และกรอง Year ตั้งแต่ปี 2002 ถึง 2016
valid_publishers = ["Nintendo", "Ubisoft", "Electronic Arts"]
df_cleaned = df[
    (df["Publisher"].isin(valid_publishers)) & (df["Year"].between(2002, 2016))
]

# 4. ลบคอลัมน์ที่ไม่จำเป็นออก (Rank, Name, Genre)
df_cleaned = df_cleaned.drop(columns=["Rank", "Name", "Genre"])

# 5. บันทึกผลลัพธ์เป็นไฟล์ excel ใหม่
df_cleaned.to_excel("Game_data_clean.xlsx", index=False)

# แสดงตัวอย่างข้อมูลที่คลีนเสร็จแล้ว
print(df_cleaned.head())
```
